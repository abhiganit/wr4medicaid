```python
import numpy as np
import pandas as pd
import geopandas as gpd
import matplotlib as mpl
import matplotlib.pyplot as plt
import pickle

```

```python
# Read US states shape file
statesmap = gpd.read_file('shapefile/tl_2017_us_state.shp')
statesmap = statesmap.to_crs(epsg=2163)

# Select rows to adjust geometries.
al = statesmap.NAME=='Alaska'
hi = statesmap.NAME=='Hawaii'
pr = statesmap.NAME=='Puerto Rico'
# adjust geometries
statesmap[al] = statesmap[al].set_geometry(statesmap[al].
                                           scale(.2,.2,.2).translate(1.4e6,-4.5e6))
statesmap[hi] = statesmap[hi].set_geometry(statesmap[hi].
                                           scale(.8,.8,.8).translate(5e6,-1.35e6))
statesmap[pr] = statesmap[pr].set_geometry(statesmap[pr].
                                           scale(2,2,2).translate(-2.3e6,0.3e6))

```

```python
statesmap.set_index('NAME',inplace=True)
```

```python
# ages- 55 or 64?
# choose ae, rr
# choose expansion vs all
upper_age_limit = 55
expansion = True

#ae = 0.524635722
#rr = 0.18
ae = 0
rr = 0
```

```python
# Read the dictionary from the pickle file
filename = f"state_ed.pkl{upper_age_limit}"
#filename = "state_ed.pkl" + str(upper_age_limit)
with open(filename, 'rb') as file:
    state_ed = pickle.load(file)
```

```python
state = 'California'

df = state_ed[state]
df
```

```python
# Add data
# https://www.census.gov/data/datasets/time-series/demo/popest/2020s-state-detail.html
# provides state level data by single age-group. So I need to add that data to the statemap for weighing

states = list(state_ed.keys())
for state in states:
    df = state_ed[state]
    df[df.columns] = df[df.columns].apply(pd.to_numeric)
    statesmap.loc[state,'Annual excess deaths (expansion)'] = df['median'].loc[(df['ae'] == ae) & (df['rr'] == rr)].iloc[0]
    statesmap.loc[state,'Annual excess deaths (all)'] = df['median'].loc[(df['ae'] == ae) & (df['rr'] == rr)].iloc[-1]
    statesmap.loc[state,'per capita excess deaths (expansion)'] = 100000*df['per_capita'].loc[(df['ae'] == ae) & (df['rr'] == rr)].iloc[0]
    statesmap.loc[state,'per capita excess deaths (all)'] = 100000*df['per_capita'].loc[(df['ae'] == ae) & (df['rr'] == rr)].iloc[-1]
```

```python
statesmap['Annual excess deaths (expansion)'].sum()
```

```python
statesmap['Annual excess deaths (all)'].sum()
```

```python
statesmap['Distribution of excess deaths (expansion)'] = statesmap['Annual excess deaths (expansion)']/statesmap['Annual excess deaths (expansion)'].sum()

statesmap['Distribution of excess deaths (all)'] = statesmap['Annual excess deaths (all)']/statesmap['Annual excess deaths (all)'].sum()
```
